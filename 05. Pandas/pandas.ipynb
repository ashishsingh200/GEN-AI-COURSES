# Pandas

```python
import pandas as pd

# from dict
df = pd.DataFrame({
    'name'    : ['Ashish', 'Prince', 'Bob', 'Alice'],
    'age'     : [20, 22, 21, 23],
    'marks'   : [85, 90, 75, 88],
    'city'    : ['Delhi', 'Mumbai', 'Pune', 'Chennai']
})

print(df)
```

```python
# from list of dicts
data = [
    {'name': 'Ashish', 'marks': 85},
    {'name': 'Prince', 'marks': 90},
]
df = pd.DataFrame(data)
print(df)
```

```python
# from list of lists
data = [['Ashish', 85], ['Prince', 90]]
df   = pd.DataFrame(data, columns=['name', 'marks'])
print(df)
```
