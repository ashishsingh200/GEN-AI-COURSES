```python
class Product: # defining a class Product having attributes name price categofy
    def __init__(self,name,price,category,discount=0): # for extra question making discount atttributes as well and defaulting it to 0
        self.name=name
        self.price=price
        self.category=category
        self.discount=discount
    
    def get_info(self): #creating get_info method  to print the product details
        print(f'The name of the product is {self.name} its price is {self.price} comes under category {self.category}.')

    def apply_discount(self): # creating another method called apply discount to print the discount amount as stated in extra question
        if self.discount !=0:
            sum=0
            sum= self.price - (self.price*self.discount)/100
            print(f'After getting a discount of {self.discount} you need to pay {sum}')
        else:
            print(f'No Discount you have to pay full price {self.price}')
```

```python
# creating 2 objects namely obj1 and obj2
obj1=Product('pen',400,'Stationary')
obj2=Product('Mac',72800,'Elevtronics',15)
```

```python
# as stated in the question applying get_info method and also to solve the extra question, calling apply_discount method
obj1.get_info()
obj1.apply_discount()
```

```python
class Product:
    def __init__(self,name,price,category):
        self.name=name
        self._price=price
        self.category=category
    
    def get_info(self): #creating get_info method  to print the product details
        print(f'The name of the product is {self.name} its price is {self._price} comes under category {self.category}.')

    #getter method
    def get_price(self):
        print(f'The price is {self._price}')

    #setter method
    def set_price(self,new_price):
        if new_price > 0:
            self._price=new_price

        else:
            print(f'The price of a given item cannot be in negative.')
        
```

```python
class ElectronicProduct(Product):
    def __init__(self, name, price, category,waranty_years, discount=0):
        super().__init__(name, price, category, discount)
        self.waranty_years=waranty_years
    
    def get_info(self):
        super().get_info()
        print(f'It has a discount of {self.discount}% and a waranty period of {self.waranty_years} years.')
```

```python
obj3=ElectronicProduct(
    name="Samsung Galaxy", 
    price=45000, 
    category="Smartphone", 
    waranty_years=2
)
obj3.get_info()
obj3.apply_discount()
```

```python
class Laptop(Product):
    def __init__(self, name, price, category):
        super().__init__(name, price, category)
    def get_info(self):
        print(f'The {self.name} laptop comes under category {self.category} priced at {self._price}')

class  Mobile(Product):
    def __init__(self, name, price, category):
        super().__init__(name, price, category)
    
    def get_info(self):
        print(f'This is  {self.name} mobile phone, comes under category {self.category} priced at {self._price}')
```

```python
from abc import ABC, abstractmethod

class Payment(ABC): #creating an abstract class payment with abstract methods

    @abstractmethod
    def process_payment(self,amount):
        pass

class CreditCardPayment(Payment):
    def process_payment(self, amount):
        self.amount=amount
        print(f"This is a credit Pay of {self.amount}")

class UPIPayment(Payment):
    def process_payment(self, amount):
        self.amount=amount
        print(f"This is a UPI payment of {self.amount}")
```

```python
cp=CreditCardPayment() # creating objects and trying all classes
```

```python
cp.process_payment(4000)
```

```python
upi=UPIPayment()
```

```python
upi.process_payment(700)
```

```python
class Product:
    def __init__(self,name, price, category):
        self.name=name
        self.price=price
        self.category=category

    def __str__(self):
        return f"{self.name} | ₹{self.price} | {self.category}"

    def __add__(self, other):
        return Product(
            self.name + " & " + other.name , 
            self.price + other.price , 
            'combined'
            )

class Inventory:
    def __init__(self):
        self.products=[]    #empty list to store products

    def add_product(self,product):
        self.products.append(product)
        print(f"{product.name} added!")

    def remove_product(self,name):
        for p in self.products:
            if p.name.lower()==name.lower():
                self.products.remove(p)
                print(f"{name} deleted.")
                return
        print(f'{p} not found!')

    def get_total_value(self):
        total=0
        for p in self.products:
            total +=p.price
        return total

    def show_all_products(self):
        if len(self.products)==0:
            print('No Products to show')
            return
        for i,p in enumerate(self.products):
            print(f'{i+1}. {p}')

class Store:
    def __init__(self,store_name):
        self.store_name=store_name
        self.inventory=Inventory()

    def add_new_product(self):
        name = input("Enter the name of the product: ")
        price=int(input("Enter the price of the product: "))
        category=input("Enter the category of the product: ")

        product=Product(name,price,category)    #creates product object
        self.inventory.add_product(product)

    def show_summary(self):
        print(f'-----{self.store_name}-----')
        print(f'Total items: {len(self.inventory.products)}')
        print(f'Total Value: {self.inventory.get_total_value()}')
```

```python
#Testing
#creating a store object
ashish=Store('Ashish_Store')

#Adding 3 products
p1 = Product("Macbook",     71648, "Electronics")
p2 = Product("Pen",         20,    "Stationery")
p3 = Product("Shampoo",     349,   "Personal Care")

ashish.inventory.add_product(p1)
ashish.inventory.add_product(p2)
ashish.inventory.add_product(p3)


#show all products
print("\n--- All Products ---")
ashish.inventory.show_all_products()

# 3. show summary
ashish.show_summary()

# 4. __add__ — combine prices of two products
combined = p1 + p2
print(f"\nCombined: {combined}")

# remove a product
ashish.inventory.remove_product("Pen")
print("\n--- After removing Pen ---")
ashish.inventory.show_all_products()
ashish.show_summary()
```
