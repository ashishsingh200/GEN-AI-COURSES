```python
import pandas as pd
```

```python
marks=pd.Series([78,85,90,66,72]) #01-->creating pandas series from list

#02-->Printing series values
print("Series Values:",marks.values)
print("Index:" ,marks.index)
print("Data Type:", marks.dtype)

#05--> Accessing the first element and last two element
print("First Element:",marks[0])
print("Last two Element:\n",marks[-2:])
```

![Task 1](images/task2.png)

```python
#01-->Adding 5 grace marks to all students
new_marks=marks.add(5)

#02-->Subtracting 2 marks from all values
sub_mask=marks.sub(2)

#03-->Multiplying all marks by 1.05
mul_marks=marks.mul(1.05)

#04-->Dividing all marks by 2
div_marks=marks.divmod(2)

#printing the result for each operation

print(f"Adding 5 grace marks to all students: {new_marks}")
print(f"Subtracting 2 marks to all students: {sub_mask}")
print(f"Multiplying by 1.05 to marks for all students: {mul_marks}")
print(f"Dividing all marks by 2 to all students: {div_marks}")
```

![Task 1](images/task3.png)

```python
#Python functions on series
#01 (a)-->Finding maximum marks
max_marks=marks.max()

#(b)-->Finding minimum marks
min_marks=marks.min()

#(c)-->Finding sum of marks
total_marks=marks.sum()

#(d)-->Mean marks
mean_marks=marks.mean()

print(f'Maximum marks : {max_marks}')
print(f'Minimum marks : {min_marks}')
print(f'Total marks   : {total_marks}')
print(f'Mean marks    : {mean_marks:.2f}')

#02-->Applying lambda function to check wether each student has passed(>=70)
lam_func = marks.apply(lambda x: 'Pass' if x >= 70 else 'Fail')  
print(f'\nPass/Fail status:\n{lam_func}')

# 03 — count students who passed
passed = lam_func[lam_func == 'Pass'].count()                      
failed = lam_func[lam_func == 'Fail'].count()                      
print(f'\nStudents passed : {passed}')
print(f'Students failed : {failed}')

```

```python
#Creating a dataframe
import pandas as pd
students={
    'Name':['Amit','Neha','Rahul','Sneha','Pooja'],
    'Marks':[78,85,90,66,72],
    'Subject':['Math','Math','Science','Science','Math']
}
#01-->Converting into data frame
df=pd.DataFrame(students)

#02-->Printing first 3 rows
print(df.head(2))

#03-->Printing last 2 rows
print(df.tail(2))

#04-->printing dataframe shape and column names
print(f'The shape of the dataframe is {df.shape}')
print(f'Column Names:\n {df.columns.tolist()}')
```

![Task 1](images/task5.png)

```python
#01(a)
info=df.info()

print(f'Info of the DataFrame\n: {info}')

#01(b)
print(f'Description of the DataFrame\n: {df.describe()}')

#01(c)
print(f"Finding first 5 data from start:\n{df.head()}")

#01(d)
print(f"Finding last 5 data from start:\n{df.tail()}")

#02-->Sorting students by marks in descendin order
sorted_stud=df.sort_values('Marks',ascending=False)
print("Students sorted by marks (highest to lowest):")
print(sorted_stud)

#03-->reseting inddex
sorted_stud.reset_index(drop=True)

#
```

```python
#01-->Student who scored more than 75 marks
above_75=df[df['Marks']>75]
print(f'Student who got more than 75 Marks: \n{above_75}')
#02-->Student who belong to subject math
math_stud=df[df['Subject']=='Math']
print(f'Student who belongs to math subject: \n{math_stud}')

#03-->Student who scored more tha average matrks
more_than_average=df[df['Marks']>df['Marks'].mean()]
print(f'Student who scored more than average marks: \n{math_stud}')

#04-->Students who failed
failed_students=df[df['Marks']<70]
print(f"The students who failed inthe exam:\n{failed_students}")
```

![Task 1](images/task7.png)

```python
#01-->finding average marks per subject using groupby()
new=df.groupby('Subject')['Marks'].mean()
print(f'Average marks per subbject:\n{new}')

#02-->Count number of students per subject
count=df.groupby('Subject')['Marks'].count()
print(f'Count of students per subject:\n{count}')

#03-->finding max marks per subject
max_marks=df.groupby('Subject')['Marks'].max()
print(f'Maximum marks per subject:\n{max_marks}')
```

```python
sales=pd.DataFrame({
    'Day':['Mon','Tue','Wed','Thu','Fri'],
    'Revenue':[1200,1500,900,2000,1800]
})

#01--->Total Revenue
total_revenue=sales['Revenue'].sum()
print(f'Total Revenue of the sale is: {total_revenue}')

#02--->Average Daily revenue
avg_rev=sales['Revenue'].mean()
print(f'Average daily Revenue of the sale is: {avg_rev}')

#03--->Day with highest revenue
high_rev_day=sales.loc[sales['Revenue']==sales['Revenue'].max(),'Day'].values[0]
print(f'Day with Highest Revenue of the sale is: {high_rev_day}')

#04--->Days where revenue is grreeater than average
day_avg=sales.loc[sales['Revenue']>sales['Revenue'].mean(),'Day'].reset_index(drop=True)
print(f'Day where Revenue of the sale is greater than average: \n{day_avg}')

#05--->pllot revenue vs day
sales.plot(kind='line',x='Day',y='Revenue',title='Revenue vs Day')

```
