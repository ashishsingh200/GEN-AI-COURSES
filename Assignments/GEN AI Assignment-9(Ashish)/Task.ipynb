```python
A=np.array([10,20,30,40])
B=np.array([1,2,3,4])
```

```python
# 01. performing Addition(A+B)
C=A+B
print(C)
```

```python
# Extra using np.add to add to array

np.add(A,B)
```

```python
# 02. Performing Subtraction (A-B)

D=A-B
print(D)
```

```python
# Extra using np.subtract() to subtract the array

np.subtract(A,B)
```

```python
# 04. Performing Multiplication

E=A*B
print(E)
```

```python
# 05. Performing Division

F=A/B
print(F)
```

```python
# 06. Performing Power (A**2)
G=A**2
print(G)
```

```python
values=np.array([2,4,6,8,10])

# 01. Calculating square root of each element
#sq_root=values**0.5
sq_root=np.sqrt(values)

# 02. Exponential of each element
exp=np.exp(values)

# 03. Natural logarithm of each element
log=np.log(values)

# 04. Sum of all elements
sum_ele=np.sum(values)

# 05. Cumulative sum of elements
cum_sum=np.cumulative_sum(values)


print(f' 01. The Square root of each element in an array {values} is: {sq_root}')
print(f' 02. The Exponential of each element in an array {values} is: {exp}')
print(f' 03. The Natural Logarithmic of each element in an array {values} is: {log}')
print(f' 04. The Sum of all element in an array {values} is: {sum_ele}')
print(f' 05. The cumulative sum of each element in an array {values} is: {cum_sum}')
```
