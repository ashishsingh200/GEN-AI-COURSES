```python
values=np.array([2,4,6,8,10])

# 01. Calculating square root of each element
#sq_root=values**0.5
sq_root=np.sqrt(values)

# 02. Exponential of each element
exp=np.exp(values)

# 03. Natural logarithm of each element
log=np.log(values)

# 04. Sum of all elements
sum_ele=np.sum(values)

# 05. Cumulative sum of elements
cum_sum=np.cumulative_sum(values)


print(f' 01. The Square root of each element in an array {values} is: {sq_root}')
print(f' 02. The Exponential of each element in an array {values} is: {exp}')
print(f' 03. The Natural Logarithmic of each element in an array {values} is: {log}')
print(f' 04. The Sum of all element in an array {values} is: {sum_ele}')
print(f' 05. The cumulative sum of each element in an array {values} is: {cum_sum}')
```

```python
#initializing the array
marks=np.array([78,85,90,66,72,88,95,60])

#caculating the tasks
#01-->Mean
mean=np.mean(marks)

#02-->Median
median=np.median(marks)

#03-->Variance
variance=np.var(marks)

#04-->Standard Deviation

s_d=np.sqrt(variance) # we can also do np.std(marks)

#05-->Minimum and maximum
min_max={}
min_max['minimum']=int(np.min(marks))
min_max['maximum']=int(np.max(marks))

#06-->Range(max-min)
ranges=np.max(marks)-np.min(marks)

#Printing the results
print(f'The Mean of marks is {mean}')
print(f'The Median of marks is {median}')
print(f'The Variance of marks is {variance}')
print(f'The Standard Deviation of marks is {s_d:.4f}')
print(f'The Minimum and Maximum of marks is {min_max}')
print(f'The Range of marks is {ranges}')
```

![task6](images/task6.png)

```python
#already initialised the marks data in task5 now performing the operations

# 01-->sorting the array
print(np.sort(marks))

# 02-->finding 25 percentile
per1=np.percentile(marks,25)
print(f'25 percentile of the marks is {per1}')

# 03-->finding 50 percentile
per2=np.percentile(marks,50)
print(f'50 percentile of the marks is {per2}')

# 04-->finding 75 percentile
per3=np.percentile(marks,75)
print(f'75 percentile of the marks is {per3}')

# 05-->calculating the no. of students who scored more than average marks
count=0
avg=np.average(marks)
for mark in marks:
    if mark>avg:
        count +=1
print(f'The no. of students who scored more than average marks is: {count}')



# we could have done in one line like this as well
count = np.sum(marks > avg)
print(f'The no. of students who scored more than average marks is: {count}')
```

```python
import numpy as np

# initialising the array
sales = np.array([1200, 1500, 900, 2000, 1800, 1700, 1600])
days  = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']

# 01 — total weekly sales
weekly_sales = np.sum(sales)
print(f'Total weekly sales    : ₹{weekly_sales}')

# 02 — average daily sales
avg_sales = np.average(sales)
print(f'Average daily sales   : ₹{avg_sales:.2f}')

# 03 — highest and lowest sales day
highest_day = days[np.argmax(sales)]    # day name of highest
lowest_day  = days[np.argmin(sales)]    # day name of lowest

print(f'Highest sales day     : {highest_day} → ₹{np.max(sales)}')
print(f'Lowest sales day      : {lowest_day}  → ₹{np.min(sales)}')

# 04 --> Standard Deviation
sd_sales=np.std(sales)
print(f'Standard Deviation of sales    : ₹{sd_sales:.4f}')

#05-->Identifying days where sales are above avg
days_arr=np.array(days)
mask=sales>np.average(sales)
days_above_avg=days_arr[mask]
sales_above_avg=sales[mask]

print(f'\nDays above average sales (₹{avg_sales:.2f}):')
for day, sale in zip(days_above_avg, sales_above_avg):
    print(f'  {day} → ₹{sale}')
```
