# Task 6: Data Cleaning 

```python
df.head()
```

 here cost and sale_amount is in string but we want it in integer or float

```python
df['Cost']=df['Cost'].str.replace('$','',regex=False).astype(float)
df['Sale_Amount']=df['Sale_Amount'].str.replace('$','',regex=False).astype(float)
```

```python
df.dtypes
```

```python
# Let's handle the missing values
# --- Numerical Column with -----> mean or median
# --- Categorical Column with ---> mode or Unknown

df.isnull().sum()
```

```python
mean=df['Clicks'].mean()
df['Clicks']=df['Clicks'].fillna(mean)
```

```python
mean=df['Impressions'].mean()
df['Impressions']=df['Impressions'].fillna(mean)
```

```python
mean=df['Cost'].mean()
df['Cost']=df['Cost'].fillna(mean)
```

```python
med=df['Leads'].median()
df['Leads']=df['Leads'].fillna(med)
```

```python
mean=df['Cost'].mean()
df['Cost']=df['Cost'].fillna(mean)
```

```python
med=df['Conversions'].median()
df['Conversions']=df['Conversions'].fillna(med)
```

```python
mean=df['Conversion Rate'].mean()
df['Conversion Rate']=df['Conversion Rate'].fillna(mean)
```

```python
mean=df['Sale_Amount'].mean()
df['Sale_Amount']=df['Sale_Amount'].fillna(mean)
```

```python
# again checking if any null values left
df.isna().sum()
```

```python
# checking any duplicate rows and removing them if any
df.duplicated().sum()
```

There are no duplicated value so we are not going to drop any since there are none

```python
# Renaming columns to lowercase and snake_case. 
''' 
    Snake case (written as snake_case) is a coding convention where you 
    write text in all lowercase letters and separate every word with an 
    underscore (_).
'''

df.columns=[col.lower().replace(' ','_') for col in df.columns]
```

### Task 10: Insights & Observations

**Insight 1 — No outliers in Cost**
The boxplot for `cost` shows no extreme outliers — the data is
evenly distributed within a normal range, suggesting consistent
ad spending across campaigns.

**Insight 2 — Conversion Rate has a skewed distribution**
The histplot + KDE for `conversion_rate` shows a right-skewed
distribution — most ads have low conversion rates with a few
performing exceptionally well. This suggests only a small
percentage of ads are highly effective.

**Insight 3 — Strong positive correlation between Clicks and Conversions**
The scatterplot of `clicks` vs `conversions` shows a clear
upward trend — more clicks generally lead to more conversions.
The heatmap confirms this with a high positive correlation value.

**Insight 4 — Device type affects Cost and Conversions**
The boxplot of `device` vs `cost` and violinplot of `device` vs
`conversions` show differences across Desktop, Mobile, and Tablet.
Desktop tends to have higher costs while mobile shows different
conversion patterns — suggesting device-specific ad strategies
may improve ROI.

**Insight 5 — Data Quality Issues Found**
The raw dataset had `Cost` and `Sale_Amount` stored as strings
with `$` symbols instead of numbers, and 7 columns had missing
values (ranging from 48 to 626 nulls). The `device` column had
9 unique values due to case inconsistency (e.g. `Desktop` vs
`desktop`) — reduced to 3 true categories after lowercasing.
