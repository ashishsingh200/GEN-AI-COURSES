# Magic Methods

#### Magic Methods are also called as dunder methods because they have double underscores before and after their name and they are special methods in python that allows us to customize how object behave.

## These Methods can help us define :

## 1. Object  Initialization(__init__)
## 2. String Representation(__str__,__repr__)
## 3. Arithmatic Operation(__add__, __sub__, etc.)
## 4. Comparision Method(__eq__,__it__)
## 5. Length, indexing, iteration(__len__,__getitem__,etc)



### Why to use Magic methods

#### 1. Makes objects behave like built in types(eg, lists, integers)
#### 2. Improves Readability by allowing basic operations object1+object2
#### 3. Simpliefies code by replacing verbose function calls with natural syntax.

```python
# __add__

class Avenger:
    def __init__(self,name,strength):
        self.name=name
        self.strength=strength

    def __add__(self, other):
        return Avenger(self.name + " & " + other.name,(self.strength + other.strength))

    def __str__(self):
        return f"{self.name} with combined strength {self.strength}"
```

```python
ironman=Avenger("Ironman",80)
thor=Avenger("Thor",120)

team_strength=ironman+thor
print(team_strength)
```
