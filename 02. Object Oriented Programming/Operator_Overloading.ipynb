# Operator Overloading

## Operator Overloading allows us to define how operators(+, -, %, /) works with user defined objects

### Why is Operator Overloading Useful ?

#### Makes Custom Objects behave like built in types
#### Improves code readability by using natural syntax (object1 + object2) instead of (object1.__add__object2)

### Operator ---> + ---> Method name ---> __add__(self,other)
### Operator ---> - ---> Method Name ---> __sub__(self,other)
### Operator ---> * ---> Method Name ---> __mul__(self,other)
### Operator ---> / ---> Method Name ---> __truediv__(self,other)
### Operator ---> = ---> Method Name ---> __eq__(self,other)
### Operator ---> > ---> Method Name ---> __gt__(self,other)
### Operator ---> < ---> Method Name ---> __lt__(self,other)
### Operator ---> != ---> Method Name ---> __ne__(self,other)

```python
class Avenger:
    def __init__(self,name,power):
        self.name=name
        self.power=power
    
    def __str__(self):
        return f" Avenger: {self.name} | Power: {self.power}"
    def __add__(self,other):
        return Avenger(self.name + " & " +other.name,self.power + other.power)

    def __mul__(self, other):
        return Avenger(self.name + " & " +other.name,self.power * other.power)
```

```python
h1=Avenger("ironman",30)
h2=Avenger("THOR",50)

total=h1+h2
print(total)
```

```python
multiple=h1*h2
print(multiple)
```
