# ABSTRACTION


## Abstraction is the concept of hiding the implementation details and only showing the necessary features of the object.
### This helps in reducing programming complexity and efforts

## Abstraction allows us to work with an object with what an object does or raather than how it does it

#### How can we achieve abstraction in python

### 1. Abstract Class --> A class that can not be instantiated or that means for an abstract class. We can't create the object of this class.


### 2. Abstract Method --> Abstract method just because it can't accessed by the object of the abstract class itself that's why we have to call this abstract method in our child class and the implementation of the abstract method is not provided inside the abstract class but must be provided in the child class.

```python
from abc import ABC, abstractmethod
class veichle(ABC):

    @abstractmethod
    def start(self):
        pass

    @abstractmethod
    def stop(self):
        pass

class Car(veichle):

    def start(self):
        print("The Car Starts with an ignition key")


    def stop(self):
        print("The car Stops by Break Paddle")


class ElectricScooter(veichle):

    def start(self):
        print("Electric Scooter is going to start with a power Button")

    def stop(self):
        print("Electric scooter stops with an electric break")
```

```python
bmw=Car()
```

```python
bmw.start()
```

```python
bmw.stop()
```

```python
antorq=ElectricScooter()
```

```python
antorq.start()
```

```python
antorq.stop()
```
