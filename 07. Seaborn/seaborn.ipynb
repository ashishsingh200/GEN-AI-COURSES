# Facet Plot

### Facet plot ---> it can help us divide the plot as per the multiple categories of a categorical column, we don't want to use hue and we want to divide the plot or graph as per multiple categories of a categorical column as row wise or column wise then we can go with the facet plot

### facet plot can help us sseperate the info row wise or column wise

### we can use row or column parameter to mention that if we want to seperate the info row wise or column wise

### facet plot functionality is only possible. in the figure level plot

```python
sns.relplot(data=tips,x='total_bill',y='tip',col='sex',hue='smoker')
plt.show()
```

```python
sns.relplot(data=tips,x='total_bill',y='tip',row='sex')
plt.show()
```

```python
sns.relplot(data=tips,x='total_bill',y='tip',col='sex',kind='line')
plt.show()
```

```python
sns.relplot(data=tips,x='total_bill',y='tip',col='sex',row='smoker')
plt.show()
```

#### A KDE (Kernel Density Estimate) plot (sns.kdeplot) visualizes the probability density of a continuous variable.
 
1. It creates a smooth, continuous curve from your data points, showing where values are most concentrated—acting like a smoothed version of a histogram.

2. Core ConceptsProbability Density: The y-axis represents density, not raw counts. 

3. The total area under the curve always equals 1.Bandwidth (bw_adjust): Controls smoothness. A lower value creates a jagged curve tracking data closely; a higher value creates a smoother curve.

```python
tips.head()
```

```python
sns.histplot(data=tips,x='total_bill',hue='sex',element='step')
```

```python
sns.kdeplot(data=tips,x='tip',y='total_bill')
```

```python
sns.displot(data=tips,x='tip',col='sex',kind='hist',kde=True)
```

### A rug plot (sns.rugplot) is a compact visualization tool that draws small vertical ticks along an axis to show the exact location of individual data points.

1. It is rarely used alone because it suffers from overplotting (ticks stack on top of each other). 

2. Instead, it is almost always paired with other distributions like histograms or KDE plots to act as a marginal indicator of raw data density.

```python
sns.rugplot(data=tips,x='sex')
```

## Heatmap

### 1. In every classification ML problem statement, we use Heatmap.

### 2. Heatmap uses color shades to show the size of numbers in a matrix 2D table

### 3. Darker or lighter colors indicate higer or lower values

### 4. Super useful when we want to see correlation, missing data,  comparision, density

#### To create a heatmap on a matrix we create a pivot table and then that pivot table only we can use to create an Heatmap

#### pivot() ---> index, columns, values 

```python
gap.head()
```

```python
piv_data=gap.pivot(index='country',columns='year',values='lifeExp')
```

```python
sns.heatmap(data=piv_data,cmap='viridis' ,cbar_kws={'label': 'Life Expectancy'})
plt.show()
```

```python
gap[gap['continent']=='Asia'].pivot(index='country',columns='year',values='lifeExp')
```

```python
h_map=gap[gap['continent']=='Asia'].pivot(index='country',columns='year',values='lifeExp')
```

```python
sns.heatmap(data=h_map)
```
