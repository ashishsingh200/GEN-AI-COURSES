## Relational Plot

### Relational Plot is being used to figure out the relation between two or more Numerical Variables and it helps us out to figure out thtat on bringing the changes on numerical column, How it affects the other Numerical Column.

#### Relationship Plot ----> Bivariate Analysis
####                   ----> Numerical - Numerical
####                   ----> One Numerical Column --> Change --> how it affercts other Numerical Column



#### Line Plot ---> It's used for Time Series Analysis(wrt time how one quantity is going to vary)

#### Scatter Plot ---> It shows how two numerical variables vary or aare corelated, that means if one numerical column will vary then how it will affect the other numerical column

```python
# let's draw a graph showing how age is related with CGPA
plt.title('Age Vs CGPA')
sns.lineplot(data=df,x='Age',y='CGPA')
plt.show()
```

```python
sns.relplot(data=df,x='Age',y='CGPA',kind='line')
plt.show()
```

```python
plt.title('Age Vs CGPA')
sns.lineplot(data=df,x='Age',y='CGPA',hue='Academic Pressure')
plt.show()
```

```python
plt.title('Age Vs CGPA')
sns.lineplot(data=df,x='Age',y='CGPA',hue='Gender')
plt.show()
```

```python
plt.title('Age Vs CGPA')
sns.lineplot(data=df,x='Age',y='CGPA',hue='Gender',style='Dietary Habits')
plt.show()
```
