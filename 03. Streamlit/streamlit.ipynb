# Streamlit

###  streamlit is an open source python library used to build beautiful, interactive web applications for data science and machine learning with minimal effort.

### Why to use streamlit?

#### 1. Easy to use.
#### 2. Fast Developments
#### 3. Interactive UI elements(like Button, charts, text input, etc)
#### 4. Great for data sscience(supports pandas, matplotlib, plotly and other libraries)
#### 5. Deployment is easy
