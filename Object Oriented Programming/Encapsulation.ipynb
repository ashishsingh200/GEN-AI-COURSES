# Encapsulation

## Encapsulation is a way of wrapping data(Variables) and methods inside a capsule or a class as asingle unit.

## It restricts the direct access to some of the objects components which is a means of preventing accidental inference and misuse of the data

#### Key Concepts of Encapsulation

1. Public Members

2. Protected Members

3. Private Members

4. Getter and Setter Methods

# Public

```python
class Student:
    def __init__(self,name,age):
        self.name=name #public
        self.age=age #public

    def show(self): #public
        print(f"Name: {self.name}, Age: {self.age}")
```

```python
s=Student('Ashish',25)
print(s.name)
print(s.age)
s.show()
```

```python
s.name='Prince' #can change directly we can modify directly
s.show()
```

```python
dir(Student)
```

## Private Members --> A Private member is not accessible outside the class and just even in the derived class. In order to have these private access over private members or to bring some modification on these members we use getter and setter method.

```python
class Avenger:
    def __init__(self,name,phone_number):
        self.name=name
        self.__phone_number=phone_number #making contact details private
    def details(self):
        print(f'My name is {self.name}. I am an Avenger.If you face any problem call me {self.__phone_number}')
        
```

```python
iron_man=Avenger('Tony Stark','8328897003')
iron_man.details()
```

```python
iron_man.name
```

```python
iron_man.phone_number
```

```python
dir(iron_man)
```

## Getter --> it will be helping us out with accessing the private data

```python
class Captain:
    def __init__(self,name,address):
        self.name=name
        self.__address=address
    #getter method
    def accessing_add(self):
        print(f'My address is {self.__address}')
```

```python
captain=Captain('Captain America','New York')
captain.name
```

```python
captain.address
```

```python
captain.__address
```

```python
captain.accessing_add()
```

## Setter --> Setter Method is being used for modifying our private data

```python
class Animal:
    def __init__(self,name,sound):
        self.name=name
        self.__sound=sound

    #getter method
    def sound_info(self):
        print(f'The animal {self.name} is going to sound as {self.__sound}')  
    def sound_modify(self,new_sound):
        self.__sound=new_sound
        print(f'new sound of the animal is {self.__sound}')
```

```python
animal=Animal("Bruno",'Woof Woof')
```

```python
animal.name
```

```python
animal.__sound
```

```python
animal.sound_modify('Bark')
```

```python
animal.sound_info()
```

## Protected data is the data which is not accessible outside the classm, but the protected data is going to be accessible by the derived class or child class


### For having the protected mamber you can mention your protected member as _attribute or by using a single underscore

```python
class GrandParent:
    def __init__(self,name,locker_key):
        self.name=name
        self._locker_key=locker_key

    ## getter method
    def locker_info(self):
        print(f'The locker key info is {self._locker_key}')
```

```python
class Parent(GrandParent):
    def __init__(self, name, locker_key):
        super().__init__(name, locker_key)
    def child_reveal(self):
        print(f'The locker key information is: {self._locker_key}')
```

```python
grandpa=GrandParent("Ashish",'123456')
```

```python
grandpa.name
```

```python
grandpa.locker_key
```

```python
papa=Parent("Aisha",'23456')
```

```python
papa.child_reveal()
```

```python
parent=Parent('Amit','1234')
```

```python
parent.locker_key
```

```python
parent.child_reveal()
```
