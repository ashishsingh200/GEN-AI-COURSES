#### plt.Figure()-->figure()--->parameters-->figsize-->(len.breadth)-->figure dimension of the graph

#### plt.Figure(figsize=(12,8)
#### plt.title()--->title()-->It takes the value os the title of the graph
#### plt.xlabel()-->it labels the x axis
#### plt.ylabel()-->it labels y axis
#### plt.grid()-->it helps to bring the grid on the charts
#### plt.show()-->helps us in representing the chart
#### plt.xlim()--> x values --> [10,1000]-->[100,1000]-->[700,800]
#### plt.ylim()--> y values --> [0,1000] -->[100,100000]


```python
count_mf=data.Gender.value_counts()
x=count_mf.index
y=count_mf.values
plt.Figure(figsize=(10,8))
plt.plot(x,y)
plt.xlabel('Gender')
plt.ylabel('No. of Peoples')
plt.title('Count plot for Genders',color='blue')
plt.tight_layout()
plt.grid()
#plt.ylim([0,30000])
plt.show()
```

## Customization for a line

#### Parameters:
1. color ----> decides what's going to be the color of the line, we can use hash code as well

2. marker --> how point of intersection appears(^,@,X,etc.)

3. linestyle---> changes the style of the line('--','-'.,'-',':',etc.)

4. linewidth --->it shows the  width of the line-->1,1.5,2,.....

5. markersize --> it shows the markersize or it can change the marker size(by default its size is 6)

6. label -->label that what this specific linr is representing but is only helpful while plotting multiple lines in a single visualization


```python
x=count_mf.index
y=count_mf.values

plt.Figure(figsize=(12,12))
plt.plot(x,y,color='#85716E',marker='X',linestyle='-.',linewidth=1,markersize=8,label='Popolation line')
plt.grid()
plt.xlabel('Gender')
plt.ylabel('Total no. of Populstion')
plt.title('Count population as per gender',color='blue')
plt.legend()
plt.show()
```
