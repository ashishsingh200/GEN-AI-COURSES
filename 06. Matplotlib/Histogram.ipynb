# Histogram

### A histogram is used to show the frequency distribution of a single continuous variable (like age, income, or transaction amounts). It groups data points into continuous ranges called bins.Unlike a bar chart, a histogram does not compare separate categories side-by-side; it shows how your data is spread out.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('Student Depression Dataset.csv')
```

```python
df.head()
```

```python
df['Age'].median()
```

```python
df['Age'].min()
```

```python
df['Age'].max()
```

```python
plt.figure(figsize=(10,8))
plt.title('Distribution of Age Group of Students')
plt.xlabel("Age")
plt.ylabel("No. of Students")
plt.hist(df['Age'],bins=[15,20,25,30,35,45,50,55,60])
plt.grid()
plt.show()
```

## Logarithmic Scale

##### Logarithmic scale --> 0 t0 10 , 10 to 102, ...

```python
plt.figure(figsize=(10,6))
plt.title('Distribution of Age Group of Students')
plt.xlabel("Age")
plt.ylabel("No. of Students")
plt.hist(df['Age'],bins=[15,20,25,30,35,45,50,55,60],log=True)
plt.grid()
plt.show()
```
