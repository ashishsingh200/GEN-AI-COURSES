#### Gender --->Equal Distribution vs Academic Pressure

```python
df['Gender'].value_counts()
```

```python
Male=df[df['Gender']=='Male'].head(12354)
```

```python
Female=df[df['Gender']=='Female']
```

### concat() -->concat() function joins two dataset or multiple dataset by using concat()

```python
new_data=pd.concat([Male,Female])
```

```python
new_data['Gender'].value_counts()
```

```python
p=new_data.groupby('Gender')['Academic Pressure'].sum()
```

```python
x=p.index
y=p.values
```

```python
plt.figure(figsize=(12,8))
plt.title('Gender Based Academic Pressure')
plt.xlabel('Gender')
plt.ylabel('Total Academic Pressure')
plt.bar(x,y,width=0.4)
plt.show()
```

# Horizontal Bar chart

```python
plt.figure(figsize=(12,8))
plt.title('Gender Based Academic Pressure')
plt.ylabel('Gender')
plt.xlabel('Total Academic Pressure')
plt.barh(x,y,color='grey')
plt.show()
```

# making bar chart with tuned sales data 

```python
data=pd.read_csv('tuned_data.csv')
```

```python
data['Product'].value_counts()
```

```python
data.head(5)
```

```python
MCC=data[data['Product']=='Mint Chip Choco']
Eclairs=data[data['Product']=='Eclairs']
White_Choc =data[data['Product']=='White Choc']
```

```python
MCC=MCC.groupby('Month')['Amount'].sum()
Eclairs=Eclairs.groupby('Month')['Amount'].sum()
White_Choc=White_Choc.groupby('Month')['Amount'].sum()
```

```python
MCC.index=pd.CategoricalIndex(MCC.index,categories=['January','February','March','April','May','June','July','August','September','October','November','December'])
Eclairs.index=pd.CategoricalIndex(Eclairs.index,categories=['January','February','March','April','May','June','July','August','September','October','November','December'])
White_Choc.index=pd.CategoricalIndex(White_Choc.index,categories=['January','February','March','April','May','June','July','August','September','October','November','December'])
```

```python
MCC=MCC.sort_index()
Eclairs=Eclairs.sort_index()
White_Choc=White_Choc.sort_index()
```

```python
plt.figure(figsize=(12,8))
plt.xlabel('Month')
plt.ylabel('Ttotal Sales')
plt.title("Different Product sales over the Month")
x=np.arange(len(MCC.index))

plt.bar(x-0.3,MCC.values,label='Mint Chips Choco',width=0.3)
plt.bar(x,Eclairs.values,label='Eclairs',width=0.3)
plt.bar(x+0.3,White_Choc.values,label='White Choc',width=0.3)
plt.legend()
plt.xticks(x,MCC.index)
plt.show()
```
