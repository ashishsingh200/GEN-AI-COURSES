# Line Plot

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('Student Depression Dataset.csv')
```

```python
df.head()
```

```python
df.size
```

```python
df.shape
```

```python
df.info()
```

```python
df['Age']=df['Age'].astype('int32')
```

```python
df.info()
```

```python
df['CGPA']=df['CGPA'].astype('float32')
```

```python
df.info()
```

```python
df['Age'].head(10)
```

```python
data=df.groupby('Age')['CGPA'].mean().head(10)
```

```python
x=data.index
y=data.values
```

```python
plt.Figure(figsize=(12,8))
plt.title('CGPA Distribution as per Age')
plt.xlabel('Age')
plt.ylabel('CGPA Values')
plt.plot(x,y,color='red',markersize=7,linestyle='--',marker='.')
plt.grid()
#plt.tight_layout()

plt.show()
```
