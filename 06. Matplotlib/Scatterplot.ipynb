# Scatter Plot

#### scatter plot ---> is being used for Bivariate Analysis --->Two numerical columns -->we check that based on the changes in one numerical column values --->HOw it affects the other column value

##### For example ----> x ---> Age -->18,29,23,22,12,21,24,54,43,34
#### y ----> cholestrol level --->50,60,40,70,23,76,54,43,34,76
### Scatter Plot --> It figures out the correlation between the two Numerical Values

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('Student Depression Dataset.csv')
df.head()
```

```python
df['Age'].value_counts().head(20)
```

```python
data=df.groupby('Age')['CGPA'].mean().head(10)
```

```python
x=data.index
y=data.values
```

```python
plt.figure(figsize=(12,8))
plt.title("Age vs CGPA")
plt.xlabel("Age")
plt.ylabel('CGPA')
plt.grid()
plt.plot(x,y,marker='o')
```

```python
plt.figure(figsize=(12,8))
plt.title("Age vs CGPA")
plt.xlabel("Age")
plt.ylabel('CGPA')
plt.grid()
plt.scatter(x,y,color='black')
```
