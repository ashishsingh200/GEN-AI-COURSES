### Classification task

```python
st=pd.read_csv('data/Student Depression Dataset.csv')
```

```python
st.head()
```

```python
st.info()
```

```python
st.isnull().sum()
```

```python
st.dropna(inplace=True)
```

```python
st.columns
```

```python
# Feature Selection
st= st[[
    'Gender', 'Age', 'City', 'Profession', 'Academic Pressure',
      'CGPA', 'Study Satisfaction',
       'Sleep Duration', 'Dietary Habits', 'Work/Study Hours',
       'Financial Stress', 'Family History of Mental Illness', 'Depression'
]]
```

```python
st.info()
```

```python
## Label encoder
from sklearn.preprocessing import LabelEncoder
```

```python
le=LabelEncoder()
```

```python
st['Gender']=le.fit_transform(st['Gender'])
st['City']=le.fit_transform(st['City'])
st['Profession']=le.fit_transform(st['Profession'])
st['Sleep Duration']=le.fit_transform(st['Sleep Duration'])
st['Dietary Habits']=le.fit_transform(st['Dietary Habits'])
st['Family History of Mental Illness']=le.fit_transform(st['Family History of Mental Illness'])
```

```python
st.head()
```

```python
st.info()
```

```python
st.shape
```

```python
x=st.iloc[:,:12]
y=st['Depression']
```

```python
x.shape
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
stds=StandardScaler()
x=stds.fit_transform(x)
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.25)
```

```python
x_train.shape
```

```python
x_test.shape
```

# Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
```

```python
lr=LogisticRegression()
```

```python
lr.fit(x_train,y_train)
```

```python
y_pred=lr.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(y_test,y_pred)
```

# Decision tree Classifier

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
tree=DecisionTreeClassifier()
```

```python
tree.fit(x_train,y_train)
y_pred=tree.predict(x_test)
```

```python
accuracy_score(y_test,y_pred)
```

# Random Forest Classifier

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
rfc=RandomForestClassifier(n_estimators=200,criterion='gini')
```

```python
rfc.fit(x_train,y_train)
y_pred=rfc.predict(x_test)

accuracy_score(y_pred,y_test)
```

# Hyperparameter Tuning

# there are two ways to perform hyperparameter tuning
1. GridSearchCV()
2. RandomSearchCV()

```python
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestClassifier
```

```python
param_grid={
    'n_estimators':[100,150,200,250],
    'max_depth':[5,10,20,40,80],
    'criterion':['log_loss','entropy','gini']
}
grid=GridSearchCV(RandomForestClassifier(),param_grid,cv=5)
```

```python
model=grid.fit(x_train,y_train)
```

```python
print(grid.best_params_)
print(grid.best_score_)
```

```python
import pickle

with open('model.pkl','wb') as file:
    pickle.dump(model,file)
```
